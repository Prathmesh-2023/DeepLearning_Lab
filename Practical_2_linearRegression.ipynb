```python
import numpy as np 
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error  
```

```python
x = np.array([100, 150, 200, 250, 300])
y = np.array([20, 25, 30, 35, 40])
```

```python
x_mean = np.mean(x)
y_mean = np.mean(y)

print(f"Mean of x: {x_mean}")
print(f"Mean of y: {y_mean}")
```

```python
numerator = np.sum((x - x_mean) * (y - y_mean))
denominator = np.sum((x - x_mean) ** 2)

print(f"Numerator: {numerator}")
print(f"Denominator: {denominator}")
print(f"Calculated slope (m): {numerator / denominator}")
```

```python
intercept = y_mean - (numerator / denominator) * x_mean
print(f"Calculated intercept (b): {intercept}")
```

```python
print(f"Linear Regression Equation: y = {numerator / denominator} * x + {intercept}")
```

```python
y_pred = (numerator / denominator) * x + intercept
print(f"Predicted values: {y_pred}")
for i in range(len(x)):
    print(f"Predicted value for x={x[i]}: {y_pred[i]}")
```

```python
residuals = y - y_pred

print(f"Residuals: {residuals}")

for i in range(len(x)):
    print(f"Residual for x={x[i]}: {residuals[i]}")
    
```

```python
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)

print("\nEvaluation Metrics:")
print(f"Mean Absolute Error (MAE): {mae}")
print(f"Mean Squared Error (MSE): {mse}")
print(f"Root Mean Squared Error (RMSE): {rmse}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(x, y, color='blue', label='Actual Data')
plt.plot(x, y_pred, color='red', label='Regression Line')
plt.title('Linear Regression')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.legend()
plt.grid(True)
```
